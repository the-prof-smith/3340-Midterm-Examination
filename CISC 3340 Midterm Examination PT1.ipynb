### Question 27 of 30 (4 points, 1 point per bug)

**2007.** The Loop's power monitor checks four readings. A reading is okay only if it is at least 50 and at most 100. ECHO should also report how many readings it checked.

**Your task:** Fix the code. It has **4 bugs**.

**Expected output:**
```text
Level 80 ok: True
Level 45 ok: False
Level 90 ok: True
Level 120 ok: False
Readings checked: 4
```

````python
power_levels = [80, 45, 90, 120]
count = 0
for level in power_level:
    is_ok = level >= 50 or level <= 100
    print("Level", level, "ok:", is_ok)
    count = 1
    print("Readings checked:", count)
````

### Question 28 of 30 (4 points, 1 point per bug)
